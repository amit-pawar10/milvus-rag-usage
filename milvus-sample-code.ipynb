```python
from pymilvus import FieldSchema, DataType, Collection, CollectionSchema

id_field = FieldSchema(
    name="id",
    dtype=DataType.INT64,
    auto_id=True,
    is_primary=True,
)


source_field = FieldSchema(
    name="source",
    dtype=DataType.VARCHAR,
    max_length=255,
    is_primary=False,
)

page_field = FieldSchema(
    name="page",
    dtype=DataType.INT64
)

embedding_field = FieldSchema(
    name="embeddings",
    dtype=DataType.FLOAT_VECTOR,
    dim=1024
)


content_field = FieldSchema(
    name="content",
    dtype=DataType.VARCHAR,
    max_length=2048
)

schema = CollectionSchema(
    fields=[
        id_field,
        source_field,
        page_field,
        embedding_field,
        content_field
    ]
)
```

```python
collection = Collection(name="science", schema = schema)
```

```python
utility.list_collections()
```

```python
index_params = {
    "metric_type": "L2",
    "index_type": "HNSW",
    "params": {
        'M': 16,
        "efConstruction": 200
    }
}
```

```python
collection.create_index(field_name="embeddings", index_params=index_params)
```

```python
collection.load()
```

```python
source = [chunk.metadata["source"] for chunk in chunks]
pages = [chunk.metadata["page"] for chunk in chunks]
page_contents = [chunk.page_content for chunk in chunks]
content_embeddings = [embedding.encode(chunk.page_content) for chunk in chunks]
```

```python
collection.insert([source, pages, content_embeddings, page_contents])
```

```python
def search_milvus(query, collection, limit=3):
  query_embedding = embedding.encode(query)

  search_params = {
    "metric_type": "L2",
    "index_type": "HNSW",
    "params": {
        'M': 16,
        "efConstruction": 200
    }
  }

  results = collection.search(
      data = [query_embedding],
      anns_field = "embeddings",
      param = search_params,
      limit = limit,
      output_fields = ["source", "page", "content"]
  )

  return results
```

```python
search_milvus("What is photosynthesis", collection)
```

```python
def filter_results(results):
  data = {}
  distances=[]
  page_numbers=[]
  contents=[]
  sources=[]

  for result in results[0]:
    distances.append(result.distance)
    page_numbers.append(result.entity.get("page"))
    contents.append(result.entity.get("content"))
    sources.append(result.entity.get("source"))

  data["distances"] = distances
  data["page_numbers"] = page_numbers
  data["contents"] = contents
  data["sources"] = sources

  return data
```

```python
results = search_milvus("What is photosynthesis", collection)
```

```python
filter_results(results)
```

```python
results = search_milvus("list the types of nervous systems", collection)
filter_results(results)
```
